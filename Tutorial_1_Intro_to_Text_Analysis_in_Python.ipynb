###5. From a String to Word Lists

So far, Python sees our sentence as one string. But text analysis often requires us to work with individual words.

One simple method is `.split():`

````python
text.split()
````

Let's save the result:

````python
words = text.split()
````

A string stores text as one piece of data. But what did .split() create?

['It', 'was', 'a', 'dark', 'and', 'stormy', 'night.']

This is called a **list**. A list allows us to store multiple pieces of data together. Lists are surrounded by square brackets [ ], with individual items separated by commas.

We can also create a list ourselves and store it in a variable:

````python
words = ["It", "was", "a", "dark", "and", "stormy", "night"]
````

We can use the len() function to get the length of a list. For text analysis, this gives us a very simple way to count the number of words in a text.

````python
len(words)
````

Len() also works for strings-but it counts characters, not words.

````python
len(text)
````

We can also retrieve individual items from a list using their position, called an index.

One unusual feature of Python is that counting begins at 0:


```
words[0]   # first word
words[1]   # second word
words[2]   # third word
```


````python
words[0]
````

Try retrieving the word "dark" from the list.

We can also retrieve several items at once. This is called slicing a list.

Turning a string into a list gives us much more control over the individual words in a text.

For example:

````python
text = "It was a dark and stormy night."
words = text.lower().split()

len(words)
````

We have now performed several steps:

**original text → lowercase text → list of words → word count**

This is a very simple example of text preprocessing: transforming raw text into a form that we can analyze computationally.

###6. Counting Words with Counter

Now that we have turned our text into a list of words, we can begin to ask questions about how frequently different words occur.

Python has a useful tool called `Counter` that counts how many times each item appears in a list.

First, we need to import `Counter`:

````python
from collections import Counter
````

Python includes many useful tools that are not loaded automatically. The import command makes one of those tools available for us to use.

Now let's create a slightly longer piece of text:

````python
text = """
The night was dark and quiet.
The night was cold.
The night was very dark.
"""
````

As before, we can convert our text to lowercase and split it into a list of words:

````python
words = text.lower().split()

words
````

Now we can use Counter on our list:

````python
word_counts = Counter(words)

word_counts
````

We can use the .most_common() method to arrange our words by frequency:

````python
word_counts.most_common()
````

Or we can ask for only the five most frequent:

````python
word_counts.most_common(5)
````

his gives us our first basic frequency analysis.

**But There Is a Problem...**

Look carefully at the results.

Why does Python treat `dark` and `dark.` as two different words?

Our `.split()` method only separated the string at spaces. It did not understand that the period in "dark." is punctuation rather than part of the word.

This introduces an important issue in text analysis: before counting words, we often need to **preprocess our text.**

For now, we can remove some punctuation using .replace():

````python
clean_text = text.lower().replace(".", "").replace(",", "")
print(clean_text)
````

Then split and count again:

````python
words = clean_text.split()

word_counts = Counter(words)

word_counts.most_common()
````

Now `"dark"` and `"dark."` are counted as the same word.

One of the most common ways to do this in Python is with a **DataFrame**.

A **DataFrame** organizes data into rows and columns, much like a spreadsheet.

To work with **DataFrames**, we will use a Python package called `pandas`.

First, import pandas:

````python
import pandas as pd
````

#### Text and Metadata

Let's create a DataFrame that looks more like a small corpus:

````python
data = {
    "author": ["Author A", "Author B", "Author C", "Author D"],
    "genre": ["fiction", "fiction", "poetry", "poetry"],
    "year": [1850, 1875, 1820, 1890],
    "text": [
        "The night was dark and quiet.",
        "The morning was bright and warm.",
        "The moon appeared over the dark sea.",
        "The sea was quiet beneath the moon."
    ]
}

df = pd.DataFrame(data)

df
````

Here we have two broad kinds of information.

**Textual data** contains the language we want to analyze:

`text`

**Metadata** describes the texts:

`author, genre, year`

Metadata is especially important in corpus analysis because it allows us to compare groups of texts.

Earlier, we applied methods to a single string using `text.lower()`

We can also perform string operations on an entire column of textual data.

For example:

````python
df["text"].str.lower()
````

Notice the addition of `.str. `in this operation.

Because df["text"] contains many strings, .str tells pandas that we want to perform a string operation on each item in the column.

We can save the results in a new column:

````python

df["text_lower"] = df["text"].str.lower()

df
````

#### Creating New Measurements

We can also calculate information about each text.

For example, count the number of characters:

````python
df["character_count"] = df["text"].str.len()

df
````

Or count the number of words:

````python
df["word_count"] = df["text"].str.split().str.len()

df
````

#### Sorting Data

Once we've created measurements, we can sort our texts according to them. For example, we can sort by word count. Default sort is from smallest to largest; reverse by setting `ascending` to False.

````python
df.sort_values("word_count")
````

````python
df.sort_values("word_count", ascending=False)
````

You could ask them:

Which text in our dataset contains the most words?

This is simple, but it's getting them accustomed to the idea that rows are observations and columns are variables/features.

#### Filtering Data

Metadata allows us to select only certain texts.

For example, suppose we only want fiction:

````python
df[df["genre"] == "fiction"]
````

We can also filter numerical information:

````python
df[df["year"] > 1850]
````

This is one of the major reasons metadata matters. We can define a subset of our corpus based on the research question we're asking.

#### Comparing Groups

We can also group texts according to their metadata.

For example:

````python
df.groupby("genre")["word_count"].mean()
````

This asks Python to:

1. Group the texts according to genre.
2. Look at word_count.
3. Calculate the average for each group.

Obviously our example dataset is much too small to draw meaningful conclusions...but imagine that df contained 10,000 poems.
